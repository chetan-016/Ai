```python
import collections
import collections.abc
# Patch collections.Mapping for compatibility with frozendict 1.2 in Python 3.10+
collections.Mapping = collections.abc.Mapping

from experta import *

class StudentFacts(Fact):
    pass

class CareerExpertSystem(KnowledgeEngine):
    # Using lowercase for rule definitions to ensure case-insensitive matching
    @Rule(StudentFacts(likes='maths'), StudentFacts(likes='physics'))
    def mechanical(self):
        print("Suggested Career Path: Mechanical Engineering")

    @Rule(StudentFacts(likes='programming'), StudentFacts(likes='maths'))
    def computer(self):
        print("Suggested Career Path: Computer Engineering")

    @Rule(StudentFacts(likes='biology'), StudentFacts(likes='chemistry'))
    def biotech(self):
        print("Suggested Career Path: Biotechnology")

    @Rule(StudentFacts(likes='circuits'), StudentFacts(likes='maths'))
    def electronics(self):
        print("Suggested Career Path: Electronics Engineering")

def main():
    engine = CareerExpertSystem()
    engine.reset()
    print("Welcome to the Career Path Expert System!")
    print("Available subjects: Maths, Physics, Programming, Biology, Chemistry, Circuits\n")

    interests = input("Enter your interests separated by commas (e.g., Maths, Physics): ").split(',')

    print("\n--- Processing Facts ---")
    for interest in interests:
        # Clean up whitespace and convert to lowercase for uniform comparison
        clean_interest = interest.strip().lower()
        if clean_interest:
            print(f"Declared interest: {clean_interest}")
            engine.declare(StudentFacts(likes=clean_interest))

    print("\n--- Results ---")
    engine.run()

if __name__ == "__main__":
    main()
```
