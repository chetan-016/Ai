```python
import collections
import collections.abc
# Patch collections.Mapping for compatibility with frozendict 1.2 in Python 3.10+
collections.Mapping = collections.abc.Mapping

from experta import *

class StudentFacts(Fact):
    pass

class CareerExpertSystem(KnowledgeEngine):
    # Track if any rule was triggered
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.triggered = False

    @Rule(StudentFacts(likes='maths'), StudentFacts(likes='physics'))
    def mechanical(self):
        self.triggered = True
        print("\n⭐ Suggested Career Path: Mechanical Engineering")

    @Rule(StudentFacts(likes='programming'), StudentFacts(likes='maths'))
    def computer(self):
        self.triggered = True
        print("\n⭐ Suggested Career Path: Computer Engineering")

    @Rule(StudentFacts(likes='biology'), StudentFacts(likes='chemistry'))
    def biotech(self):
        self.triggered = True
        print("\n⭐ Suggested Career Path: Biotechnology")

    @Rule(StudentFacts(likes='circuits'), StudentFacts(likes='maths'))
    def electronics(self):
        self.triggered = True
        print("\n⭐ Suggested Career Path: Electronics Engineering")

def main():
    engine = CareerExpertSystem()
    engine.reset()

    # Mapping common abbreviations to matching rule words
    alias_map = {
        'math': 'maths',
        'mathematics': 'maths',
        'coding': 'programming',
        'code': 'programming',
        'bio': 'biology',
        'chem': 'chemistry',
        'circuit': 'circuits'
    }

    print("=== Career Path Expert System ===")
    print("Rules require a combination of two subjects (e.g., maths + physics, programming + maths, biology + chemistry, circuits + maths)")
    print("Available subjects: Maths, Physics, Programming, Biology, Chemistry, Circuits\n")

    user_input = input("Enter your interests separated by commas (e.g., math, physics): ")
    interests = user_input.split(',')

    print("\n--- Processing Facts ---")
    declared_any = False
    for interest in interests:
        clean_interest = interest.strip().lower()
        if clean_interest:
            # Translate alias if found (e.g., 'math' -> 'maths')
            resolved_interest = alias_map.get(clean_interest, clean_interest)
            if resolved_interest != clean_interest:
                print(f"Declared interest: {clean_interest} (mapped to '{resolved_interest}')")
            else:
                print(f"Declared interest: {clean_interest}")

            engine.declare(StudentFacts(likes=resolved_interest))
            declared_any = True

    if not declared_any:
        print("No interests entered.")
        return

    print("\n--- Results ---")
    engine.run()

    if not engine.triggered:
        print("No career path matched. Try entering a matching pair like 'maths, physics' or 'programming, maths'!")

if __name__ == "__main__":
    main()
```
